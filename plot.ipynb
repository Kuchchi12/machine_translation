```python
import numpy as np
import matplotlib.pyplot as plt

d_model = 512.0
warmup_steps = 4000.0
steps = np.arange(1, 20001)
arg1 = steps ** -0.5
arg2 = steps * (warmup_steps ** -1.5)
lr = (d_model ** -0.5) * np.minimum(arg1, arg2)

plt.plot(steps, lr)
plt.xlabel('Training Steps')
plt.ylabel('Learning Rate')
plt.title('Transformer LR Schedule (d_model=512, warmup=4000)')
plt.axvline(x=4000, color='r', linestyle='--', label='Warmup End')
plt.legend()
plt.show()
```
